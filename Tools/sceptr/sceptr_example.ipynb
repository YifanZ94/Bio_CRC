```python
from pandas import DataFrame
import sceptr
tcrs = DataFrame(
    data = {
            "CDR2A": ["CAHRSAGGGTSYGKLTF", "CAVDNARLMF", "CAERIRKGQVLTGGGNKLTF", "CAYRSAGGGTSYGKLTF", "CAYRSAGGGTSYGKLTF"],
            "CDR2B": ["CASSEFQGDNEQFF", "CASSDGSFNEQFF", "CASSVGDLLTGELFF", "CASSPGTGGNEQYF", "CASSPGTGGNEQYF"],
        
            "TRAV": ["TRAV38-1*01", "TRAV3*01", "TRAV13-2*01", "TRAV38-2/DV8*01", "TRAV38-2/DV8*01"],
            "CDR3A": ["", "CAVDNARLMF", "CAERIRKGQVLTGGGNKLTF", "CAYRSAGGGTSYGKLTF", "CAYRSAGGGTSYGKLTF"],
            "TRBV": ["TRBV2*01", "TRBV25-1*01", "TRBV9*01", "TRBV2*01", "TRBV2*01"],
            "CDR3B": ["CASSEFQGDNEQFF", "CASSDGSFNEQFF", "CASSVGDLLTGELFF", "CASSPGTGGNEQYF", "CASSPGTGGNEQYF"],
    },
    index = [0,1,2,3,4]
)
```

```python
### model variants
from sceptr import variant
sceptr_tiny = variant.tiny()
tiny_reps = sceptr_tiny.calc_vector_representations(tcrs)
print(tiny_reps.shape)
```

```python
### cross-distance matrix between two sets of TCR
cdist_matrix = sceptr.calc_cdist_matrix(tcrs.iloc[:2], tcrs.iloc[2:])
print(cdist_matrix)
```

```python
### within-set distances
pdist_vector = sceptr.calc_pdist_vector(tcrs)
print(pdist_vector)
## return shape 0.5*N*(N-1)
```

```python
### directly operate on sceptr’s TCR representations
reps = sceptr.calc_vector_representations(tcrs)
reps.shape
```
