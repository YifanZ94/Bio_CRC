```python
import hashlib, json, os
from datetime import datetime

def hash_file(path):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            h.update(chunk)
    return h.hexdigest()

manifest = {
    "file": DATA_PATH,
    "md5": hash_file(DATA_PATH),
    "size_mb": round(os.path.getsize(DATA_PATH) / 1e6, 2),
    "timestamp": datetime.now().isoformat(),
    "note": "init md hashing"
}

with open("data_manifest.json", "w") as f:
    json.dump(manifest, f, indent=2)
```

```python
def check_manifest(manifest_path="data_manifest.json"):
    with open(manifest_path) as f:
        manifest = json.load(f)

    file_path = manifest["file"]
    
    # Check file still exists
    if not os.path.exists(file_path):
        print(f"❌ FILE MISSING: {file_path}")
        return False

    # Re-hash and compare
    current_md5 = hash_file(file_path)
    current_size = round(os.path.getsize(file_path) / 1e6, 2)

    print(f"File     : {file_path}")
    print(f"Recorded : {manifest['md5']}  ({manifest['size_mb']} MB)  @ {manifest['timestamp']}")
    print(f"Current  : {current_md5}  ({current_size} MB)  @ {datetime.now().isoformat()}")

    if current_md5 == manifest["md5"]:
        print("✅ File unchanged — matches manifest")
        return True
    else:
        print("⚠️  File has changed since manifest was recorded!")
        return False

check_manifest("data_manifest.json")
```
