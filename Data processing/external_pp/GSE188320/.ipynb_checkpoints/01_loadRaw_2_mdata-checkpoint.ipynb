# Load Batch 6-9 Data from GSE188320

This notebook loads GEX and TCR data from batches 6-9 in the GSE188320 dataset, combining them into a MuData object with batch information stored in obs metadata.


```python
import scanpy as sc
import scirpy as ir
import pandas as pd
import mudata as mu
import scipy.io as spio
import shutil
from pathlib import Path

sc.settings.verbosity = 3
```

```python
gex_files = ["GSM5677747_batch6_hashed_EAE_mouse1_filtered_",
      "GSM5677750_batch6_hashed_EAE_mouse2_filtered_",
      "GSM5677753_batch7_hashed_EAE_mouse1_filtered_",
      "GSM5677756_batch7_hashed_EAE_mouse2_filtered_",
      "GSM5677759_batch8_hashed_UT_mouse1_filtered_",
      "GSM5677762_batch8_hashed_UT_mouse2_filtered_",
      "GSM5677765_batch9_hashed_UT_mouse1_filtered_",
      "GSM5677768_batch9_hashed_UT_mouse2_filtered_",]

tcr_files = [
    "GSM5677749_batch6_hashed_EAE_mouse1_",
    "GSM5677752_batch6_hashed_EAE_mouse2_",
    "GSM5677755_batch7_hashed_EAE_mouse1_",
    "GSM5677758_batch7_hashed_EAE_mouse2_",
    "GSM5677761_batch8_hashed_UT_mouse1_",
    "GSM5677764_batch8_hashed_UT_mouse2_",
    "GSM5677767_batch9_hashed_UT_mouse1_",
    "GSM5677770_batch9_hashed_UT_mouse2_",
]

meta_files = [
    "GSM5677748_batch6_hashed_EAE_mouse1_",
    "GSM5677751_batch6_hashed_EAE_mouse2_",
    "GSM5677754_batch7_hashed_EAE_mouse1_",
    "GSM5677757_batch7_hashed_EAE_mouse2_",
    "GSM5677760_batch8_hashed_UT_mouse1_",
    "GSM5677763_batch8_hashed_UT_mouse2_",
    "GSM5677766_batch9_hashed_UT_mouse1_",
    "GSM5677769_batch9_hashed_UT_mouse2_",
]
```

```python
# Loop through all samples and combine into MuData
import os
import gzip

gex_data = []
tcr_data = []
metadata_data = []

for i in range(len(meta_files)):
    # load GEX
    adata_ext = sc.read_10x_mtx(path='.', var_names='gene_symbols', 
                            make_unique=True, cache=True, gex_only=True,
                            prefix = gex_files[i])
    
    # Add metadata to obs
    adata_ext.obs['batch'] = gex_files[i].split('_')[1]
    adata_ext.obs['condition'] = gex_files[i].split('_')[3]
    adata_ext.obs['mouse_id'] = gex_files[i].split('_')[4]
    sub_fix = '-b' + gex_files[i].split('_')[1][-1] + 'm' + gex_files[i].split('_')[4][-1]
    adata_ext.obs.index = adata_ext.obs.index + sub_fix

    
    # Load TCR data
    print("  Loading TCR...")
    # Extract the .gz file
    with gzip.open(tcr_files[i] + "all_contig_annotations.csv.gz", 'rb') as f_in:
        with open(tcr_files[i] + "all_contig_annotations.csv", 'wb') as f_out:
            shutil.copyfileobj(f_in, f_out)
    
    adata_tcr = ir.io.read_10x_vdj(tcr_files[i] + "all_contig_annotations.csv")
    adata_tcr.obs.index = adata_tcr.obs.index + sub_fix    
    tcr_data.append(adata_tcr)


    # load meta data
    with gzip.open(meta_files[i] + "metadata_demuxEM.csv.gz", 'rb') as f_in:
        with open(meta_files[i] + "metadata_demuxEM.csv", 'wb') as f_out:
            shutil.copyfileobj(f_in, f_out)

    meta_L = pd.read_csv(meta_files[i] + "metadata_demuxEM.csv", on_bad_lines='skip')
    if isinstance(meta_L.index, pd.MultiIndex):
        meta_L = meta_L.reset_index()
        meta_L = meta_L.drop(columns=['level_1'])  
        meta_L = meta_L.rename(columns={'level_0': 'BC'})
        meta_L = meta_L.set_index('BC')
        cols = meta_L.columns.tolist()
        meta_L.columns = cols[1:] + ['is_multiplet']

    meta_L = meta_L.dropna(subset=['assignment_demuxem'])
    # Drop rows where assignment_demuxem is a number (either int/float or string representation of a number)
    mask = ~meta_L['assignment_demuxem'].astype(str).str.fullmatch(r'\d+(\.\d+)?')
    meta_L = meta_L[mask]
    meta_L = meta_L[meta_L['assignment_demuxem'] != 'Other']
    
    meta_L.index = meta_L.index + '-1' + sub_fix
    
    # Merge assignment_demuxem into adata_ext obs by intersecting indices
    intersect_indices = adata_ext.obs.index.intersection(meta_L.index)
    adata_ext.obs.loc[intersect_indices, 'assignment_demuxem'] = meta_L.loc[intersect_indices, 'assignment_demuxem']

    gex_data.append(adata_ext)
```

```python
# Combine gex_data and tcr_data with intersect index
gex_combined = sc.concat(gex_data, join='outer')
tcr_combined = sc.concat(tcr_data, join='outer')

# Find intersect and create MuData
common_cells = gex_combined.obs.index.intersection(tcr_combined.obs.index)
mdata = mu.MuData({
    "gex": gex_combined[common_cells],
    "airr": tcr_combined[common_cells]
})

```
